# Exploratory Data Analysis
This notebook focuses on fraud rate, amount differences, and time-based patterns.

```python
from pathlib import Path
import pandas as pd

path = Path('../data/raw/creditcard.csv')
if path.exists():
    df = pd.read_csv(path)
    print('Fraud rate:', df['Class'].mean())
    print(df.groupby('Class')['Amount'].describe())
else:
    print('Dataset not available yet. Place the Kaggle CSV file in data/raw/creditcard.csv.')
```
